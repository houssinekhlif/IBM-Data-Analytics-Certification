#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here
cols=["Employment","JobSat","RemoteWork"]
na_perc=[]
print("Missing Values Percentage by Column:\n")
for col in cols:
    perc=round(df[col].isna().sum()/df.shape[0],2)
    print(col,":",perc)
    na_perc.append(perc)
for index,col in enumerate(cols):
    if na_perc[index]>=0.6:
        df.drop(columns=col,axis=1,inplace=True)
    else:
        if df[col].dtypes=="object":
            df[col].fillna(df[col].value_counts().idxmax(),inplace=True)
        else:
            df[col].fillna(df[col].median())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here

years_map={"Less than 1 year":0,"More than 50 years":51}
df["YearsCodePro"].replace(years_map,inplace=True)
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors="coerce")
df["YearsCodePro"].fillna(df["YearsCodePro"].median(),inplace=True)
cuts=[0,5,10,20,51]
labels=["Low","Medium","High","Extremely High"]
df['ExperienceLevel']=pd.cut(df["YearsCodePro"],bins=cuts,labels=labels,right=True,include_lowest=True)
df_grouped=df.groupby("ExperienceLevel",as_index=False)["JobSat"].median()
print(df_grouped)
sns.barplot(df_grouped,x="ExperienceLevel",y="JobSat",palette='viridis')
plt.title("Job Satisfaction by ExperienceLevel/YearsCodePro")
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here
df["LanguageHaveWorkedWith"]=df["LanguageHaveWorkedWith"].str.split(";")
df=df.explode("LanguageHaveWorkedWith",ignore_index=True)
df["LanguageHaveWorkedWith"]=df["LanguageHaveWorkedWith"].str.strip()

top_countries=df["Country"].value_counts().head(15).index
top_langs=df["LanguageHaveWorkedWith"].value_counts().head(20).index
df_filtered=df[(df["LanguageHaveWorkedWith"].isin(top_langs))&(df["Country"].isin(top_countries))]

country_crosstab=pd.crosstab(df_filtered["Country"],df_filtered["LanguageHaveWorkedWith"],normalize="index")

plt.figure(figsize=(12,6))
ax=sns.heatmap(country_crosstab,cmap='coolwarm')
labels=country_crosstab.index.tolist()
labels[13:15]=["United Kingdom","United States"]
ax.set_yticklabels(labels=labels)
plt.title("Coding Language by Countries with the highest numnber of respondents")
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here
top_emp=df["Employment"].value_counts().head(10).index
df_filtered1=df[df["Employment"].isin(top_emp)]
crosstable=pd.crosstab(df_filtered1["EdLevel"],df_filtered1["Employment"])
plt.figure(figsize=(12,8))
ax=sns.heatmap(crosstable,cmap='YlGnBu',annot=True)
plt.xticks(fontsize=6)
plt.yticks(fontsize=6)
labels=["Primary/elementary school","Bachelor’s degree","Master’s degree","Some college/university","Secondary school","Professional degree","Associate degree","Something else"]
ax.set_yticklabels(labels)
plt.title("Education Level VS Employment")
plt.show()
```

```python
plt.figure(figsize=(15,7))
ax=sns.countplot(data=df_filtered1,x="EdLevel",hue="Employment")
labels=["Primary/elementary school","Bachelor’s degree","Master’s degree","Some college/university","Secondary school","Professional degree","Associate degree","Something else"]
plt.xticks(rotation=45)
ax.set_xticklabels(labels)
plt.title("Education Level VS Employment")
plt.xlabel("Education Level")
plt.show()
```
