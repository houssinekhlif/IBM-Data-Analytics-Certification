### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
df1=df.copy()
df1["DatabaseWantToWorkWith"]=df1["DatabaseWantToWorkWith"].str.split(";")
df1=df1.explode("DatabaseWantToWorkWith")
df1["DatabaseWantToWorkWith"]=df1["DatabaseWantToWorkWith"].str.strip()
dbs_pie=df1["DatabaseWantToWorkWith"].value_counts().head()

plt.figure(figsize=(8,8))
colors=['#8dd3c7', '#ffffb3', '#bebada', '#fb8072', '#80b1d3'] 
plt.pie(dbs_pie.values,
        labels=dbs_pie.index,
        colors=colors,
        autopct="%.2f%%",
        wedgeprops={'edgecolor': 'gray', 'linewidth': .5})
plt.axis("equal")
plt.title("Top 5 Databases Respondents Want To Work With")
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
dev_pie=df["DevType"].value_counts().head()
plt.figure(figsize=(8,8))
colors=['#8dd3c7', '#ffffb3', '#bebada', '#fb8072', '#80b1d3'] 
plt.pie(dev_pie.values,
        colors=colors,
        autopct="%.2f%%",
        wedgeprops={'edgecolor': 'gray', 'linewidth': .5})
plt.axis("equal")
plt.title("Top 5 Developer Roles",fontsize=18,fontweight="bold")
plt.legend(dev_pie.index,bbox_to_anchor=[0.8,0.9],title="Developer Type")
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
##Write your code here
df2=df.copy()
df2["OpSysProfessional use"]=df2["OpSysProfessional use"].str.split(";")
df2=df2.explode("OpSysProfessional use")
df2["OpSysProfessional use"]=df2["OpSysProfessional use"].str.strip()
opsys_pie=df2["OpSysProfessional use"].value_counts().head(10)

plt.figure(figsize=(8,8))
colors=plt.cm.tab20(range(len(opsys_pie)))
plt.pie(opsys_pie.values,
        colors=colors,
        autopct="%.2f%%",
        pctdistance=.85,
        wedgeprops={'edgecolor': 'gray', 'linewidth': .5})
plt.axis("equal")
plt.title("Top 10 Operating Systems Used By Respondents For Professional Use")
plt.legend(opsys_pie.index,bbox_to_anchor=[1,1],title="Operating System Name")
plt.show()
```

### Task 2: Additional Visualizations and Comparisons


##### 2.1 Pie Chart for Top 5 Programming Languages Respondents Have Worked With


The `LanguageHaveWorkedWith` column contains the programming languages that respondents have experience with. We’ll plot a pie chart to display the composition of the top 5 languages.



```python
##Write your code here
df3=df.copy()
df3["LanguageHaveWorkedWith"]=df3["LanguageHaveWorkedWith"].str.split(";")
df3=df3.explode("LanguageHaveWorkedWith")
df3["LanguageHaveWorkedWith"]=df3["LanguageHaveWorkedWith"].str.strip()
lang_pie=df3["LanguageHaveWorkedWith"].value_counts().head()

plt.figure(figsize=(8,8))
colors=plt.cm.tab20(range(len(lang_pie)))
plt.pie(lang_pie.values,
        colors=colors,
        autopct="%.2f%%",
        pctdistance=.85,
        startangle=90,
        wedgeprops={'edgecolor': 'gray', 'linewidth': .5})
plt.axis("equal")
plt.title("Top 5 Programming Languages Respondents Have Worked With")
plt.legend(lang_pie.index,bbox_to_anchor=[1,1],title="Programming Languague Name")
plt.show()
```

##### 2.2 Pie Chart for Top Collaboration Tools used in Professional Use


Using the `NEWCollabToolsHaveWorkedWith` column, we’ll identify and visualize the top collaboration tools respondents use in their professional work.



```python
##Write your code here
df["NEWCollabToolsHaveWorkedWith"].value_counts()

df4=df.copy()
df4["NEWCollabToolsHaveWorkedWith"]=df4["NEWCollabToolsHaveWorkedWith"].str.split(";")
df4=df4.explode("NEWCollabToolsHaveWorkedWith")
df4["NEWCollabToolsHaveWorkedWith"]=df4["NEWCollabToolsHaveWorkedWith"].str.strip()
collab_pie=df4["NEWCollabToolsHaveWorkedWith"].value_counts().head(10)

plt.figure(figsize=(8,8))
cmap = plt.cm.get_cmap('viridis')
num_colors =len(collab_pie)
colors = [cmap(i / num_colors) for i in range(num_colors)]
plt.pie(collab_pie.values,
        colors=colors,
        autopct="%.2f%%",
        pctdistance=.85,
        startangle=90,
        wedgeprops={'edgecolor': 'gray', 'linewidth': .5})
plt.axis("equal")
plt.title("Top 10 Collaboration Tools Respondents Use In Their Professional Work")
plt.legend(collab_pie.index,bbox_to_anchor=[1,1],title="Collaboration Tool Name")
plt.show()
```
